```python
#using the format function
first_name = 'Rose'
second_name = 'Mary'
print('Hello I am {} {}'.format(first_name,second_name))
```

```python
print('Hello I am {0} {1}'.format(first_name,second_name))
```

```python
print('Hello I am {0} {1}, {1} is my surname'.format(first_name,second_name))
```

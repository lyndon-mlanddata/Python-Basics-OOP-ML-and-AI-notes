# for and while Loops

### for loop
iterate over a sequence like a list dictionary or tuple

```python
a = ['Delhi','Mumbai','New Tork', 'Birmingham']
for city in a: #you are creating a nw variable "city" to use 
    print(city)
```

```python
numbers = [1,2,3,4,5,6,7,8]
for i in numbers:
    if(i%2!=0):
        print (i, ' is not divisible by 2')
        
```

### while Loop
iterates over a block of code as long as the condition is true 

```python
x = 1
while x<=5:
    print(x)
    x = x+1
else: #tip: theres acctually no need for a else statment here
    print("i is no longer less than 5")
```

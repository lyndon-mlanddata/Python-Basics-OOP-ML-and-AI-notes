# continue and break Loop

### break statemnt
used to break out of a for or while loop

```python
for i in range(1,10):
    print(i)
    if i ==5:
        break
#kinda like a while loop but made in a diferent way
```

### continue statement
skips the current iteration of the loop and goes to the next

```python
for i in range(1,10):
    if i ==5:
        continue
    print(i)
```

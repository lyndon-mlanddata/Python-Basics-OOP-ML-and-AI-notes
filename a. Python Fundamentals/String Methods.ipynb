# String Methods

#### capitalise() -  returns the string after convertign the first character into uppercase

```python
s = 'bollywoood'
s.capitalize()
```

#### lower() - which lreturns the string after lowering everything into lowercase

```python
t = "TERMINATOR"
t.lower()
```

#### upper() - returns the string after converting all letters into uppercase

```python
v = "vector"
v.upper()
```

#### join() - returns the string after concantenating elements of the string with a separator

```python
q1 = ["Lyndon", "Brendan", "Paul" ]
q = "@@".join(q1)
print(q)
```

#### len() - returns the length of the string

```python
string_len = "You are me"
print(len(string_len))
#it counts the spaces, so
#the spaces count as characters as well
```

#### count() - function which returns the number of occurences of the substring

```python
st = "Lionel Messi"
st.count("e")
```

#### find() - returns the index of the substring or else returns -1

```python
st1 = "Ronaldo"
st1.find("o")
```

```python
st1.find("t")
```

#### islower() - checks if all characters of the string are lower case

```python
lcs = "Wet"
lcs.islower()
```

#### islower() - checks if all characters of the string are upper case

```python
ucs = "OMAR"
ucs.isupper()
```

#### split(" ") - returns a list of the string after breaking the string by the sepaator

```python
spl = "Naruto+Sasuke+Sakura"
spl.split("+")
```

#### strip() - removes any extra spaces at the beginning and at the end of the string

```python
stri = "    Why are you here?"
stri.strip()
```

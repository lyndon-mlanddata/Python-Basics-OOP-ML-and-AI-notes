# Operations on lists
### append()
adds an element at the end of a list

```python
s = [ "apple", "cherry"]
s.append("banana")
print(s)
```

### insert() 
adds an elembnt at the specified index

```python
s.insert(1,"papaya")
```

```python
print(s)
```

### remove()
removes the specified element

```python
s.remove("banana")
```

```python
print(s)
```

### pop()
removes the element at the specified index or the last item if the element is not specified

```python
t=["grape","orange","banana","kiwi","banana"]
t.pop() # without an index it removes the last item
```

```python
t.pop(3) # with an index 
print(t)
```

### copy()
returns a copy of the list

```python
tcopy = t.copy()
print(tcopy)
```

### clear()
removes all elements form list

```python
u = [ 1,2,3,4,5,6]
u.clear()
print(u)
```

### reverse()
reverses the order of the elements in a list

```python
t.reverse()
print(t)
```

### count()
returns the number of elements wiht a specified value

```python
t.count("apple") # there are 0 apples so 0
```

```python
t.count("banana") # there is 1 banana
```

### extend()
adds the elemnts of another list to the current list

```python
s1 = [1,2,3,4,5,6,7]
s2 =["a","b","c"]
s1.extend(s2)
print(s1)
```

### sort()
sorts the list, alphabetical order

```python
e = [2,5,8,3,1,57,78,24,7,3,]
e.sort()
print(e)
```

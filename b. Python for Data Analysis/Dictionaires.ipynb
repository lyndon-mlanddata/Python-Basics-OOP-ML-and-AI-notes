# Dictionaries
An unordered collection of daya values, used to store data values, in the key:value pair

- the value in the dictionary can repeat and can be of any data type
- the keys of the dictionary must be of immutable types(eg string, integer, float)

Dictionaies are written with {}

Typical dictionary looks like
- {key1:value1,key2:value2...}

```python
d = {"Brand" : "Ford", "Model":"Mustang","Year":1964}
```

```python
print(d)
```

### Creating a dictionary using the dict function
dict function dictionaries look like
- dict(key1=value1,key2=value2...)
- there is no need to have "" around keys in this method

```python
thisdict = dict(Brand = "Ford", Model="Mustang",Year=1964)
print(thisdict)
```

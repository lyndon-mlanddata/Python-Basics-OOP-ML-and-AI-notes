# Implementing Stacks and Queues using Dequeue 

- Dequeue stands for "Double Ended Queue"

### Implementing Stacks using Dequeue

```python
#importing th edeque library
from collections import deque
```

```python
numbers = deque()
# adding items into deque
numbers.append(23)
numbers.append(45)
numbers.append(12)
numbers.append(54)
numbers
```

```python
numbers.pop()
```

```python
numbers.pop()
```

### Implement a Queue using Deque 

```python
names = deque()
```

```python
names.append("Naruto")
names.append("Sasuke")
names.append("Kakashi")
names.append("Itachi")
names
```

```python
#we can use popleft which is a functio of deque
names.popleft()
```

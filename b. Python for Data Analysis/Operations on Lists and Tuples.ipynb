# Operations on Tuples and Lists

### count()
returns the number of times a specified value occurs in a tuple

```python
t = ("apple", "banana", "cherry","banana")
t.count("banana")
```

### index()
returns the postion of a specified item

```python
t.index("cherry") #2 meaning the third element
```

### len()
returns the length of the tuplesb

```python
len(t)
```

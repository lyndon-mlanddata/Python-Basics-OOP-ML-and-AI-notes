## Linear Search

```python
my_list = [2,3,5,4,7,8,9]B
```

```python
# searching the list using for loop
for i in range(0,len(my_list)): 
    if my_list[i] = 8:
        print(i)
    else:
        print('element not found')
# worst case scenario would be 
# if we have to find the element at the
# end of the list, that is O(n)
# O(n) simply means 
# the amount of operations performed =
# the elments in the list
```

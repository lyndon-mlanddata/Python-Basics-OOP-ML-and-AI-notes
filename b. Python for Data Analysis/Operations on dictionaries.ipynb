# Operations on Dictionaries
### copy()
returns a copy of the dictionary

```python
dict1 = {1:"Lyndon",2:"Paul",3:"Byron"}
dict2 = dict1.copy()
print(dict2)
```

### items()
returns a list containing a tuple for each key:value pair

```python
dict1.items()
```

### keys() 
returns a list containing the dictionary keys

```python
dict1.keys()
```

### pop()
removes the element with the specified key

```python
dict1.pop(1)
dict1
```

### popitem()
removes the last inserted key/value pair

```python
dict1.popitem()
```

```python
dict1
```

### update()
updates the dictionary with a new specified key/value pair

```python
dict1.update({4:"Naruto"})
dict1
```

### clear()
removes all items from the dictionary

```python
dict1.clear()
dict1
```

### values()
returns a list of all values in the dictiionary

```python
dict2 = {1:"Lyndon2",2:"Paul2",3:"Byron2"}
dict2.values()
```

# Set Operations

2. Intersection
3. Difference

### union() or |
the set of all elements from bnth sets

```python
A = {1,2,4,5,6,7,9}
B = {3,6,9,12,15}
print(A|B)
print(A.union(B))
```

### intersection() or &
the set of elements that are present in both sets

```python
print(A&B)
print(A.intersection(B))
```

### difference or  - 
set of elemetns that are only in A but not in B

```python
A = {1,2,4,7,8}
B = {1,2,3,5,9,8}
A-B
A.difference(B)
#1,2,8 are in both A and B
#3,5,9 are in B
# that leaves 4 and 7 that are ONLY in A
```

```python
B-A
B.difference(A)
```

### ^ or symmetric_difference()

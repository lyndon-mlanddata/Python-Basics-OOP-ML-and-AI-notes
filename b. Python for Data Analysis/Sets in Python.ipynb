# Sets in Python
- a set is an unordered/unindexed collection of items
- they're written with curly brackets {}

```python
s = {"apple","cherry","banana"}
```

```python
for i in s:
    print(i)
```

### add()
- adds items into a set

```python
s.add("pineapple")
s
```

### update()
- adds multiple items into the set

```python
s ={"apple","cherry","banana"}
s.update({"watermelon","orange"})
s
```

### remove() 
- removes elements in the set that are in the set

### discard()
- removes items that are in the set but doesnt raise an error if that item is not present in the set

```python
s.discard("apple")
s
```

```python
s.discard("papaya")
# no error
```

### pop()
- removes the last element of the set, however since the elements of a set have no order or ibndex, the pop function can remove any element 

```python
s.pop()
s
```

```python
s.clear()
s
```

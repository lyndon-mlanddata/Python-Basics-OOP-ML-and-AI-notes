# Nested Dictionaires 
- Dictionaires inside dictrionaires
- the look like dictwhole = {dict1{key1:value1,key1.5:value1.5}, dict2{key2:value2}}
- the second dictionary is stored in the key of value of the first dictionary

```python
student_marks = { "John":{"Math":34,"English":89,"Science":67},
"Callum":{"Math":98,"English":80,"Science":92},
"Lyndon":{"Math":70,"English":40,"Science":68}}

student_marks
```

- accesing the marks scored by "Callum"

```python
student_marks["Callum"]
```

- accesing the marks scored by "Callum" in "Maths"

```python
student_marks["Callum"]["Math"]
```

- adding a new student

```python
student_marks["Millie"] = {}
student_marks["Millie"]["Math"] = 99
student_marks["Millie"]["English"] = 98
student_marks["Millie"]["Science"] = 88

student_marks
```

- removing items from the nested dictionary

```python
student_marks.pop("John")
student_marks
```

- iterating through a nested dictionary

```python
student_marks.items()
```

```python
for name,subjectsandMarks in student_marks.items():    
    print("Name of student: ", name)
    for subjects in subjectsandMarks:
        print(subjects,":",subjectsandMarks[subjects])
```

- line 1 - we are naming the name of the student "name" and he subjects and marks "subjectsandMarks" in the list
this is because student_marks.items() creates a list containing tuples of each key:value pair (see above)
- line 2 -we print the name of the student 
- line 3 and 4 - "subjectsandMarks" contains stuff like "Math" : 90, "english" :68 etc so this for loop goes in and says for subjects in 
subjectsandMarks print the subjects and then print the marks next to the subjects by doing
subjectsandMarks[subjects] we print the marks next to the subjects, eg in "Math" : 90 (subjectsandMarks) we do subjectsandMarks[subjects]
which is 90 '''


the nested dictionary still acts as a dictionary as shown by the syntax subjectsandMarks[subjects]

### Implementing Stacks Using Lists

```python
# empty list
letters = []
```

```python
#adding elements
# thisis a PUSH operation - adds the element and that elemt is the last to be removed
letters.append('a')
letters.append('e')
letters.append('f')
letters.append('h')
letters
```

```python
#removing elements
#pop operation - removes the last element
letters.pop()
```

```python
letters.pop()
```

```python
# implementig queue operation
fruits.pop(0)
```

# Insertion and Selection

- Insertion works by taking an element out of the unsorted list and 
putting it in a new list in the correct order

- Selection works by taking the smallest element in a list and putting it at the
beginning of that unsorted list , after that, the lsit is separated into 2 parts and we repeat the process , we then repeat the process again for subsequent unsorted lists

they both have the same time complexity of O(n^2)

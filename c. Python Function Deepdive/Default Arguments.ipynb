# Default parameters in Functions 

```python
def myFunc(a,b): #a and b are parameters
    return a-b
```

```python
myFunc(2,3) #the values 2 and 3 are arguments
```

```python
def myFunc2(a,b=4):
    return a*b
```

```python
myFunc2(2) # no need to enter b's value because it is already defined as 4
```

```python
myFunc2(2,7) # however we can still change the value of b by just putting any value we want
```

```python
def myFunc3(a,b=4,c=7): 
    # c cannot be allowed to not have a 
    # default value because it wont work, everytbing after b  must have a default value
    return a*b-c
```

```python
myFunc3(3)
```

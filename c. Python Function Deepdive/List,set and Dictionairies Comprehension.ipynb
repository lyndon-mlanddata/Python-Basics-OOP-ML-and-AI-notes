# Comprehensions in Python
These provide us with a new way of constructing new sequences and lists , set and dictionaries, using sequences which have already been defined\

### List Comprehension

```python
#defining a list
my_list = [2,3,4,5,6]
#using list comprehension
squared_list = [val**2 for val in my_list] # shorter syntax instead of for loop
squared_list
```

```python
# using list comprehension
even_list = [var for var in my_list if var%2 == 0]
even_list
```

### Set Comprehension

```python
# using set comprehension to create a set of numbers 1 to 9
numbers = {x for x in range(1,10)}
numbers
```

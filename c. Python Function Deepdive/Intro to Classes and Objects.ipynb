# Introduction to Classes and Objects
An object is a collection of data and methods that act upon that data

*Class* is a blueprint for objects (think biology: classes are like genotype and the object is the phenotype)
*Object* is an instance of the class

```python
class person:
    age=20
```

```python
Nick=person()
```

```python
class person: 
    def __init__(self,name,age):
        self.name = name
        self.age = age
        # self is used to access variables of the class
        # always define the self parameter FIRST
```

```python
p1 = person("Nathan",32)
print(p1.name)
print(p1.age)
```

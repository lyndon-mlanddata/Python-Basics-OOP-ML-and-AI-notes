# date class
Three manditory attributes (year, month , day)
There is a range to all the attributes

- Minimum year <= year <= Maximum year
- 1 <= month <= 12
- 1 <= day <= nummber of days in the given month

```python
from datetime import date
mydate = date(year=2045,month=4,day=18)
print(mydate)
date.today()
```

# time class
- 4 attributes
- hour, minute second and microsecond

```python
from datetime import time
mytime = time(hour=14,minute=45,second=59,microsecond=90)
print(mytime)
```

# timedelta class
this is used to manipulate difference in dates, and it is used for date manipulation in python

```python
# date and time exactly 3 months from now
from datetime import datetime, timedelta
todays_date = datetime.now()
dateafter_month = todays_date+timedelta(days=90)
dateafter_month
```

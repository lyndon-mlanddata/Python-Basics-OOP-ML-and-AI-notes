# Aggregate Functions
Aggregation Functions a re functions where multiple values are grouped together to form a single summary value 

Common aggregate fucntions are
1. max()
2. min()
3. sum()
4. mean()
5. median()
6. mode()

```python
list1 = [3,5,7,9,2,1,1]
max(list1)
```

```python
min(list1)
```

```python
sum(list1)
```

```python
import statistics as s
```

```python
s.mean(list1)
```

```python
s.median(list1)
```

```python
s.mode(list1)
```

# Keyword Arguments

```python
def totalMarks(m,s,e):
    return m+s+e
totalMarks(23,56,89)
```

```python
totalMarks(m=23,s=56,e=89) #another way to call up the arguments
```

```python
totalMarks(s=23,56,89) #we must define the other variables, else we get an error
```

```python
def marks(**score):
    print("marks in maths " + score["maths"])

marks(science = '67',english = '89', maths = '78')
```

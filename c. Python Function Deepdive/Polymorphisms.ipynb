- we will be exploring the different ways of using different methods

```python
# 3 BANKS AND THEIR RATE OF INTEREST
class SBI:
    def rate_of_interest(self):
        return 8

class ICICI:
    def rate_of_interest(self):
        return 4

class HDFC:
    def rate_of_interest(self):
        return 7

```

```python
def common_interest(bank):
    return bank.rate_of_interest()
```

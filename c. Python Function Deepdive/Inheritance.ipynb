# Inheritance

```python
class person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def display(self):
        print(self.name,self.age)
p1 = person("Naruto", 15)
p1.display() #call the display function with display()
```

```python
class student(person):
    def __init__(self,name,age,grade):
        super().__init__(name,age)# inherit the attributes from the person class by sending 
        self.grade = grade
    def studInfo(self):
        if self.grade <=10:
            print("your student cannot be younger than this")
        else:
            print(self.name,self.age,self.grade)
        # the person class as a parameter while defining studnet class
s1 = student("Kira",21,8)
s1.display()
s1.studInfo()
```

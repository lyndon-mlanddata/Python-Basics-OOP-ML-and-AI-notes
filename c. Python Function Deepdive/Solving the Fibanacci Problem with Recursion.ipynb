# Solving the Fibonacci Problem using Recursion
- Fibonnacci Sequence is a sequence where each number is the sum of the preceeding one
-  for example 0,1,1,2,3,5,8,13,21 etc

```python
def fibonacci(n): # n is the (n + 1) th term in the fibonnacci sequence eg n = 6 is the 7th term in the fibonacci sequence
    if n<=1:                                 
        return n
    else:
        return fibonacci(n-1)+fibonacci(n-2) # f(n) = f(n-1) + f(n-2)
fibonacci(6)
```

```python
def func(n):
    if n == 6:
        return n
    else:
        return 2*func(n+1)
func(5)
```

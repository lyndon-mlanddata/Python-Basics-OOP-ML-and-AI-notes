# Python Modules
A Python Modules is a file with python code, it may define python functions or lists etc

Lets make a "Simple Math" Module below

```python
# we will describe four simpple functions that we will be able to use in this module

#addition
def add(a,b):
    return a+b

#subtraction
def subt(a,b):
    return a-b

#multiplication
def mult(a,b):
    return a*b

#division
def div(a,b):
    return a-b
```

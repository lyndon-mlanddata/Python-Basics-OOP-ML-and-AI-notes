## Textual Analysis

You might also want to analyse some sentiment in a piece of text that the data has, this section will go over what you need to do for textual analysis of the data

```python
# In the case of reviews, we might want to check the length of reviews and the size of the ratings and whetehr there is a pattenr between them
# Checking whether Length of Review has any Impact on Ratings of the Drugs

# for that we need to create a new column to calculate length of the reviews
data['len']  = data['columnWithReview'].apply(len)
# check the Impact of Length of Reviews on another column
data[['columnOfChoiceToCompareWithText','len']].groupby(['columnOfChoiceToCompareWithText']).agg(['min','mean','max'])
```

```python
# we will remove unnecassry things such as Stopwords, Punctuations, numbers etc

# First lets remove Punctuations from the Text we are trying ot analyse
def punctuation_removal(messy_str):
    clean_list = [char for char in messy_str if char not in string.punctuation]
    clean_str = ''.join(clean_list)
    return clean_str

data['columnWithText'] = data['columnWithText'].apply(punctuation_removal)
```

```python
#Removing the stopwords form the Text we are trying to analyse
# Now lets Remove the Stopwords also

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

stop = stopwords.words('english')
stop.append("i'm") #  you can add or remove words from the stopwords list if you
# depending on your research and analysis goals

stop_words = []

# Removing punctuation form stopwords
for item in stop:
    new_item = punctuation_removal(item) # removing the punctuation from each word in the stopwords
    stop_words.append(new_item) #add into the new stopwords list


def stopwords_removal(messy_str):
    messy_str = word_tokenize(messy_str)
    return [word.lower() for word in messy_str
            if word.lower() not in stop_words ]

data['columnWithText'] = data['columnWithText'].apply(stopwords_removal)
```

```python
# lets remove the Numbers also

import re
def drop_numbers(list_text):
    # list_text will contain our text with each word
    list_text_new = []
    for i in list_text:
        if not re.search('\d', i):
            list_text_new.append(i)
    return ' '.join(list_text_new)

data['columnWithText'] = data['columnWithText'].apply(drop_numbers)
```

Sentiment analysis on the text

```python
# for using Sentiment Analyzer we will have to dowload the Vader Lexicon from NLTK

import nltk
nltk.download('vader_lexicon')
"""this basically contains expressions in english that
convey different emotions, we will use the vader lexicon for sentimental analysis
(evaluate emotion) in text in our text of choice"""

# lets calculate the Sentiment from our text of chocie

from nltk.sentiment.vader import SentimentIntensityAnalyzer
sid = SentimentIntensityAnalyzer()

train_sentiments = []

for i in data['columnWithText']:
    train_sentiments.append(sid.polarity_scores(i).get('compound'))
    # get("compund") gives the overall sentiment in a piece of text
train_sentiments = np.asarray(train_sentiments)
data['sentiment'] = pd.Series(data=train_sentiments) # adding a new column of sentiment

# lets check Impact of Sentiment on Reviews
data[['columnOfChoiceToCompareWithText','sentiment']].groupby(['rating']).agg(['min','mean','max'])

"""the code above will plot the ratings against the min,mean,max of all the sentiments of the reviews
that were rated that number, for example, for a raitng 1 with 34 reviews, it will plot rating 1
against the min,mean,max of all those 34 reviews"""
```
